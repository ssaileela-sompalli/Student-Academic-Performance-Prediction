```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
import pandas as pd

df = pd.read_csv(
    "student_dataset/data/student-mat.csv",
    sep=";"
)

X = df.drop("G3", axis=1)
y = df["G3"]

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)
print("Input Features:", X.shape)
print("Target Variable:", y.shape)
```

```python
X_encoded = pd.get_dummies(X, drop_first=True, dtype=int)

print("Original Features Shape:", X.shape)
print("Encoded Features Shape:", X_encoded.shape)

print("\nFirst 5 Encoded Records:")
display(X_encoded.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

```python
from flaml import AutoML

automl = AutoML()

automl.fit(
    X_train=X_train,
    y_train=y_train,
    task="regression",
    time_budget=60,
    metric="rmse",
    eval_method="cv",
    n_splits=5,
    seed=42
)

print("Best Model:", automl.best_estimator)
print("Best RMSE:", automl.best_loss)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

y_pred = automl.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("MODEL EVALUATION RESULTS")
print("------------------------")
print("MAE:", round(mae, 4))
print("MSE:", round(mse, 4))
print("RMSE:", round(rmse, 4))
print("R2 Score:", round(r2, 4))
```

```python
comparison = pd.DataFrame({
    "Actual Grade": y_test.values,
    "Predicted Grade": np.round(y_pred, 2)
})

comparison["Prediction Error"] = (
    comparison["Actual Grade"] - comparison["Predicted Grade"]
)

display(comparison.head(15))
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred, alpha=0.7)

plt.plot(
    [0, 20],
    [0, 20],
    color="red",
    linestyle="--",
    label="Perfect Prediction"
)

plt.xlabel("Actual Grades")
plt.ylabel("Predicted Grades")
plt.title("Actual vs Predicted Student Grades")
plt.legend()
plt.grid(True)

plt.show()
```

```python
from sklearn.inspection import permutation_importance

importance = permutation_importance(
    automl,
    X_test,
    y_test,
    n_repeats=10,
    random_state=42,
    scoring="r2"
)

feature_importance = pd.DataFrame({
    "Feature": X_test.columns,
    "Importance": importance.importances_mean
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

display(feature_importance.head(10))
```

```python
top_features = feature_importance.head(10).sort_values(
    by="Importance",
    ascending=True
)

plt.figure(figsize=(9, 6))

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.xlabel("Permutation Importance")
plt.ylabel("Features")
plt.title("Top 10 Important Features for Student Grade Prediction")
plt.grid(axis="x", alpha=0.3)

plt.show()
```

```python
# Select one student from the test dataset
student = X_test.iloc[[0]]

# Predict the student's final grade
predicted_grade = automl.predict(student)

print("Predicted Final Grade:", round(predicted_grade[0], 2))
```

```python
grade = predicted_grade[0]

if grade < 10:
    category = "Needs Improvement"
elif grade < 15:
    category = "Average"
elif grade < 18:
    category = "Good"
else:
    category = "Excellent"

print("Predicted Grade:", round(grade, 2))
print("Performance Category:", category)
```

```python
def predict_student_performance(student_data):

    predicted_grade = automl.predict(student_data)[0]

    if predicted_grade < 10:
        category = "Needs Improvement"
    elif predicted_grade < 15:
        category = "Average"
    elif predicted_grade < 18:
        category = "Good"
    else:
        category = "Excellent"

    print("Predicted Final Grade:", round(predicted_grade, 2))
    print("Performance Category:", category)

    return predicted_grade, category


# Test the function
predict_student_performance(X_test.iloc[[0]])
```

```python
# Remove G1 and G2 from the input features
X_without_grades = df.drop(
    ["G1", "G2", "G3"],
    axis=1
)

# One-hot encode categorical columns
X_without_grades_encoded = pd.get_dummies(
    X_without_grades,
    drop_first=True,
    dtype=int
)

# Display the new dataset shape
print("New Feature Shape:", X_without_grades_encoded.shape)
print("Target Shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train_new, X_test_new, y_train_new, y_test_new = train_test_split(
    X_without_grades_encoded,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Data:", X_train_new.shape)
print("Testing Data:", X_test_new.shape)
print("Training Target:", y_train_new.shape)
print("Testing Target:", y_test_new.shape)
```

```python
from flaml import AutoML

automl_new = AutoML()

automl_new.fit(
    X_train=X_train_new,
    y_train=y_train_new,
    task="regression",
    time_budget=60,
    metric="rmse",
    eval_method="cv",
    n_splits=5,
    seed=42
)

print("Best Model:", automl_new.best_estimator)
print("Best CV RMSE:", automl_new.best_loss)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

y_pred_new = automl_new.predict(X_test_new)

mae_new = mean_absolute_error(y_test_new, y_pred_new)
mse_new = mean_squared_error(y_test_new, y_pred_new)
rmse_new = np.sqrt(mse_new)
r2_new = r2_score(y_test_new, y_pred_new)

print("SECOND MODEL EVALUATION")
print("-----------------------")
print("MAE:", round(mae_new, 4))
print("MSE:", round(mse_new, 4))
print("RMSE:", round(rmse_new, 4))
print("R2 Score:", round(r2_new, 4))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

comparison = pd.DataFrame({
    "Model": ["With G1 and G2", "Without G1 and G2"],
    "MAE": [mae, mae_new],
    "RMSE": [rmse, rmse_new],
    "R2 Score": [r2, r2_new]
})

display(comparison)

comparison.set_index("Model")[["RMSE"]].plot(
    kind="bar",
    figsize=(7, 4),
    legend=False
)

plt.title("RMSE Comparison of Both Models")
plt.ylabel("RMSE (Lower is Better)")
plt.xticks(rotation=0)
plt.show()
```

```python
from sklearn.inspection import permutation_importance

importance_new = permutation_importance(
    automl_new,
    X_test_new,
    y_test_new,
    n_repeats=10,
    random_state=42,
    scoring="r2"
)

importance_df_new = pd.DataFrame({
    "Feature": X_test_new.columns,
    "Importance": importance_new.importances_mean
}).sort_values(by="Importance", ascending=False)

display(importance_df_new.head(10))
```

```python
top_features = importance_df_new.head(10).sort_values(
    by="Importance",
    ascending=True
)

plt.figure(figsize=(9, 5))

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.xlabel("Permutation Importance")
plt.ylabel("Student Features")
plt.title("Top 10 Important Features - Model Without G1 and G2")
plt.tight_layout()
plt.show()
```

```python
import joblib

model = joblib.load(
    "student_performance_model_without_grades.pkl"
)

print("Student Performance Model Loaded Successfully!")
```

```python
def predict_student(
    school, sex, age, address, famsize, Pstatus,
    Medu, Fedu, Mjob, Fjob, reason, guardian,
    traveltime, studytime, failures, schoolsup,
    famsup, paid, activities, nursery, higher,
    internet, romantic, famrel, freetime, goout,
    Dalc, Walc, health, absences
):
    student = pd.DataFrame([{
        "school": school,
        "sex": sex,
        "age": age,
        "address": address,
        "famsize": famsize,
        "Pstatus": Pstatus,
        "Medu": Medu,
        "Fedu": Fedu,
        "Mjob": Mjob,
        "Fjob": Fjob,
        "reason": reason,
        "guardian": guardian,
        "traveltime": traveltime,
        "studytime": studytime,
        "failures": failures,
        "schoolsup": schoolsup,
        "famsup": famsup,
        "paid": paid,
        "activities": activities,
        "nursery": nursery,
        "higher": higher,
        "internet": internet,
        "romantic": romantic,
        "famrel": famrel,
        "freetime": freetime,
        "goout": goout,
        "Dalc": Dalc,
        "Walc": Walc,
        "health": health,
        "absences": absences
    }])

    student_encoded = pd.get_dummies(
        student, drop_first=True, dtype=int
    )

    student_encoded = student_encoded.reindex(
        columns=X_without_grades_encoded.columns,
        fill_value=0
    )

    prediction = model.predict(student_encoded)[0]
    prediction = max(0, min(20, prediction))

    return round(float(prediction), 2)
```
